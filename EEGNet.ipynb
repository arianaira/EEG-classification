```python
import pandas as pd
import mat73
import numpy as np
```

```python
problem_object = []
for i in range(1,8) :
    file_path = f'/content/d{i}.mat'
    data_dict = mat73.loadmat(file_path)


    variable_name = f'd{i}'

    problem_object.append(data_dict[variable_name])
```

```python
data = np.concatenate(problem_object)
data.shape
```

```python
CONTROL = range(1,10129)
ADD = range(10129,23160)
ADHD = range(23160,33903)
```

```python
def set_type(id) :
    if id in CONTROL :
        return 0
    if id in ADD :
        return 1
    if id in ADHD :
        return 2
```

```python
full_dataset_X = data
full_dataset_y = [set_type(idd) for idd in range(1,33903)]
```

```python
np.unique(full_dataset_y)
```

```python
full_dataset_X = np.array(full_dataset_X)
full_dataset_y = np.array(full_dataset_y)
```

```python
full_dataset_X.shape
```

```python
# np.save('/content/Full_dataset_X.npy' , full_dataset_X)
# np.save('/content/Full_dataset_y.npy' , full_dataset_y)
```

```python
# Create the balanced X dataset correctly
balanced_dataset_X = np.concatenate([
    full_dataset_X[:10000],
    full_dataset_X[10129:20129],
    full_dataset_X[23160:33160]
])

# Create the balanced y dataset correctly using np.concatenate
balanced_dataset_y = np.concatenate([
    full_dataset_y[:10000],
    full_dataset_y[10129:20129],
    full_dataset_y[23160:33160]
])
```

```python
# --- CRITICAL STEP: Free up RAM ---
# After creating the new balanced arrays, delete the large original ones.
import gc
del full_dataset_X
del full_dataset_y
gc.collect() # Ask the garbage collector to release the memory

# Now you can safely work with your balanced dataset
print(f"Shape of balanced X: {balanced_dataset_X.shape}")
print(f"Shape of balanced y: {balanced_dataset_y.shape}")

# There is no need for these lines, as the arrays are already np.array
# balanced_dataset_X = np.array(balanced_dataset_X)
# balanced_dataset_y = np.array(balanced_dataset_y)
```

```python
import numpy as np

trial_means = balanced_dataset_X.mean(axis=-1)  # mean per trial per channel
trial_stds  = balanced_dataset_X.std(axis=-1)   # std per trial per channel

print("Channel means (first trial):", trial_means[0])
print("Channel stds (first trial):", trial_stds[0])
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from EEGModels import EEGNet

X_train, X_test, y_train, y_test = train_test_split(
    balanced_dataset_X,
    balanced_dataset_y,
    test_size=0.2,  # 20% for testing
    random_state=42,
    stratify=balanced_dataset_y
)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,  # 20% of training for validation
    random_state=42,
    stratify=y_train
)
```

```python
del balanced_dataset_X
gc.collect()
```

```python
def zscore_normalize(X):
    # X shape: (trials, channels, samples)
    mean = X.mean(axis=-1, keepdims=True)
    std = X.std(axis=-1, keepdims=True)
    return (X - mean) / (std + 1e-6)  # add epsilon to avoid /0

X_train = zscore_normalize(X_train)
X_val   = zscore_normalize(X_val)
X_test  = zscore_normalize(X_test)
```

```python
num_classes = len(np.unique(balanced_dataset_y))
Y_train = to_categorical(y_train, num_classes)
Y_val   = to_categorical(y_val, num_classes)
Y_test  = to_categorical(y_test, num_classes)

n_chans   = X_train.shape[1]
n_samples = X_train.shape[2]

model = EEGNet(nb_classes=num_classes,
               Chans=n_chans,
               Samples=n_samples,
               dropoutRate=0.5)

model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

# --- 4. Train ---
history = model.fit(X_train, Y_train,
                    epochs=30,
                    batch_size=16,
                    validation_data=(X_val, Y_val),
                    verbose=1)

# --- 5. Evaluate ---
score = model.evaluate(X_test, Y_test, verbose=0)
print("Test accuracy:", score[1])
```
